```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from datetime import datetime, timedelta

# Load dataset
file_path = "D:/niteditedfinal.csv"
data = pd.read_csv(file_path)

# Features and targets
X = data[['Year', 'Month', 'Day', 'Hour', 'Minute']]
y_temp = data['Temperature']
y_ghi = data['GHI']

# Train-test split (only once)
X_train, X_test, y_train_temp, y_test_temp = train_test_split(
    X, y_temp, test_size=0.2, random_state=42
)

_, _, y_train_ghi, y_test_ghi = train_test_split(
    X, y_ghi, test_size=0.2, random_state=42
)

# Train models (only once)
rf_temp = RandomForestRegressor(random_state=42)
rf_temp.fit(X_train, y_train_temp)

rf_ghi = RandomForestRegressor(random_state=42)
rf_ghi.fit(X_train, y_train_ghi)

# Function to prepare input
def create_input(dt):
    return np.array([[dt.year, dt.month, dt.day, dt.hour, dt.minute]])

# Real-time + future prediction loop
current_time = datetime.now()

for i in range(2):   # current + next 15 min 
    
    X_new = create_input(current_time)

    prediction_temp = rf_temp.predict(X_new)[0]
    prediction_ghi = rf_ghi.predict(X_new)[0]

    if i == 0:
        print(f"Current Time: {current_time}")
        print(f"Predicted Temperature: {prediction_temp}")
        print(f"Predicted GHI: {prediction_ghi}")
    else:
        print(f"\nAfter {i*15} minutes ({current_time}):")
        print(f"Predicted Temperature: {prediction_temp}")
        print(f"Predicted GHI: {prediction_ghi}")

    # Move time forward by 15 minutes
    current_time += timedelta(minutes=15)
```
